```python
arr = np.array([1, 2, 3, 4])
print("Array a:", arr)
```

```python
print("Array:", arr)
print("Type:", type(arr)) #Python nesnesinin sınıfını gösterir
print("Data type:", arr.dtype) #Dizinin içindeki elemanların veri tipini gösterir
print("Number of dimensions:", arr.ndim) #Dizinin boyut sayısını tam sayı olarak verir.
print("Shape:", arr.shape) #Dizinin her boyuttaki eleman sayısını bir tuple olarak verir.
#Örneğin (2, 3) çıktısı dizinin 2 satır ve 3 sütundan oluştuğunu;
#tek boyutlu (5,) çıktısı ise 5 elemanlı tek satır/vektör yapısını gösterir.
print("Size (total elements):", arr.size) #Dizideki toplam eleman sayısını verir.
```

```python
data = np.array([5, 10, 15, 20, 25])
```

```python
print("Median:", np.median(data)) #Medyanı bulur
print("Standard deviation:", np.std(data)) #Verilerin ortalamadan ne kadar uzaklaştığını/yayıldığını ölçer (standart sapma)

matrix = np.arange(1, 10).reshape(3, 3) #Bu 9 elemanlı tek boyutlu diziyi 3 satır ve 3 sütunluk 2 boyutlu bir matrise dönüştürür.
print("Matrix:\n", matrix)
print("Row sums:", matrix.sum(axis=1)) #(Satır top) Sütunlar boyunca ilerleyip her satırı kendi içinde toplamak demektir.
print("Column sums:", matrix.sum(axis=0)) #(Sütun top) Satırlar boyunca aşağı inip her sütunu kendi içinde toplamak demektir.
print("Diagonal sum:", np.trace(matrix)) #(Köşegen top) Matrisin ana köşegenindeki elemanların ((0,0), (1,1), (2,2)) toplamını verir.
```

```python
arr = np.arange(12)
print("Original:", arr)
```

```python
reshaped = arr.reshape(3, 4)
print("Reshaped (3x4):\n", reshaped)
```

```python
transposed = reshaped.T
print("Transposed:\n", transposed) #Transpoze işlemi
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Vertical stack:\n", np.vstack((a, b))) #Kaç dizi verirsek o kadar satırlı bir dizi oluşturur.
print("Horizontal stack:\n", np.hstack((a, b))) #1D dizilerde elemanları yan yana uç uca ekler ve tek boyutlu bir dizi döndürür.
```

```python
x = np.arange(10)
print("Split into 2 parts:", np.split(x, 2)) #Diziyi 2 eşit parçaya böler (her biri 5 elemanlı iki dizi).
print("Split into 5 parts:", np.array_split(x, 5)) #Diziyi 5 parçaya böler (10 eleman 5'e tam bölündüğü için her biri 2 elemanlı 5 dizi üretir
```

```python
#Koşullu filtreleme
mask = arr > 5
print("Mask:", mask)
print("Values > 5:", arr[mask])
print("Values ≤ 5:", arr[~mask])
```

```python
arr = np.array([50, 60, 70, 80, 90])
```

```python
updated = np.where(arr > 75, arr + 5, arr) #75'ten büyük değerlere 5 puan ekle
print("Original:", arr)
print("Updated:", updated)
```

```python
x = np.array([1, 2, 2, 3, 3, 3])
print("Unique values:", np.unique(x)) #Tekrarsız şekilde elemanlar
print("Cumulative sum:", np.cumsum(x)) #Her adımda o ana kadarki sayıların toplamını yazar ve diziyle aynı uzunlukta bir sonuç döndürür.
print("Sort ascending:", np.sort(x)) #Dizideki elemanları küçükten büyüğe sıralayıp sıralı yeni bir dizi döndürür.
print("Sorted indices:", np.argsort(x)) #Elemanların kendisini değil, küçükten büyüğe sıralanmış hallerini oluşturmak için 
#hangi orijinal indekslerin hangi sırayla alınması gerektiğini döndürür.
```
