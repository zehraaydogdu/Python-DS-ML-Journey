# Fonksiyonlar

```python
def hello_python():
    print("Hello")
    print("Python") #Olarak tanımlanır.
```

- Parametresiz

```python
hello_python() #Fonksiyon çağrılır. Fonksiyon çağrıldığında fonksiyonun içindeki kodlar çalışır.
```

- Parametreli

```python
def hello_name(name):
    print("Hello")
    print(name) #Fonksiyon parametre alır. Parametre ile fonksiyona veri aktarılır.
```

```python
hello_name(): #Bu hata verir. Fonksiyon çağrılırken parametre verilmelidir. Çünkü fonksiyon parametre alacak şekilde tanımlanmıştır.
```

```python
hello_name("Zehra") #Fonksiyon çağrılır ve parametre ile veri aktarılır.
```

- Default yani varsayılan değer parametreli

```python
def hello_surname(surname = "Aydogdu"):
    print("Hello")
    print(surname)
```

```python
hello_surname("Yilmaz")
```

```python
hello_surname() #Parametre verilmediği için varsayılan değer (default)kullanılır.
```

- Geriye değer döndüren 

```python
def toplama(a, b, c):
    print(a + b + c)
```

```python
toplama(10, 20, 30) #Fonksiyon çağrılır ve parametreler ile veri aktarılır.
```

```python
x = toplama(10, 20, 30) #Fonksiyon çağrılır ve parametreler ile veri aktarılır. Fonksiyonun geri döndürdüğü değer x değişkenine atanır.
```

```python
type(x) #Fonksiyonun geri döndürdüğü değer NoneType'dır. Çünkü fonksiyon geriye değer döndürmez.
```

```python
def return_toplama(a, b, c):
    return a + b + c
```

```python
x = return_toplama(10, 20, 30)
```

```python
type(x) #Fonksiyonun geri döndürdüğü değer int'tir. Çünkü fonksiyon geriye değer döndürür.
```

```python
def args_toplam(*args):
    return sum(args) #Çağrılırken kaç tane parametre verilirse verilsin, hepsini bir tuple olarak alır.
#Artık istediğim kadar değer gönderebilirim. Birden fazla ya da sayısı belli olmayan input durumlarında kullanılır.
```

```python
args_toplam(10, 20, 30, 40, 50)
```

```python
def kwargs_ornek(**kwargs):
    print(kwargs)
```

```python
kwargs_ornek(elma=100, muz=200) #Bizim için bir sözlük oluşturur.
```
