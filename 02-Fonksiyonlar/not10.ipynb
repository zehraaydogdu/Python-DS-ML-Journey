## LEGB -> L:Local, E:Enclosing, G:Global, B:Built-in

```python
#Global
myString = "Zehra"

def myFunction():
    #Enclosing
    myString = "Zehra 2"
    print(myString)
    
    def myFunction2():
        #Local
        myString = "Zehra 3"
        print(myString)
    
    myFunction2()
```

```python
myString
```

- *Herkes kendi kapsamında çalışır*
*Önce Local'in içine bakar. Varsa orda çalıştırır. Eğer yoksa bunu çevreleyen yapıya bakar*
*Yoksa globalde bakar.   
Eğer yoksa pythonda tanımlı mı diye bakar (built-in)*

```python
x = 2

def example():
    global x #Global kapsamdaki değişkenlere normalde erişemiyoruz fonk içinden. Erişmek için 'global' anahtarı kullanıyoruz.
    x = 3

example()
print(x) 
```

```python
x = 5
y = 10
z = 8
def example1():
    x = 15
    y = 20
    z = 7
    def example2():
        nonlocal y #İç içe fonklarda bir dış kapsamdaki değişkene erişmek ve işlem gerçekleştirebilmek için 'nonlocal' anahtarı kullanılır.
        print(x)
        print(y)
        print(z)
    example2()
    
example1()

print(z)
```

# Generators(Üreteçler)

**Normal Fonksiyonlar**: Başlar ve tek seferde baştan sona çalışıp biter.  
**Generator Fonksiyonlar**: Çağrıldığında doğrudan gövdesini çalıştırmaz; bir generator nesnesi döndürür.  
next() çağrıldığında çalışmaya başlar ve ilk yield ifadesine kadar ilerler.   
yield gördüğünde değeri üretir ve duraklayıp bulunduğu satırı/değişken durumunu hafızada tutar.   
next() tekrar çağrıldığında kaldığı yerden devam edip bir sonraki yielde kadar çalışır.   
Üretecek başka yield kalmadığında akış Finish noktasına ulaşır.  

Büyük verileri tek seferde okuyarak belleğe yüklemek yerine, generative fonksiyonlar ile parça parça işlemek mümkündür.

```python
def example():
    #kodlar
    yield 1
    
    #kodlar
    yield 2
    
    #kodlar
    yield 3 #return fonksiyonu tek değer döndürüp fonksiyonu tamamen sonlandırırken, yield her çağrıldığında   
    #sıradaki değeri üretip fonksiyonu beklemeye alır. Bu sayede büyük verileri tümüyle RAM'e yüklemek yerine parça parça işlemek mümkün olur. 

gen = example() #Üreteci oluşturur.

print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen)) #3 tane yield var artık tükendi.
```

```python
def tek_sayilar():
    n = 1
    while n <= 19:
        yield n
        n += 2

for tek_sayi in tek_sayilar():
    print(tek_sayi)
```
