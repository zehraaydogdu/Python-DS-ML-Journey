## Hazır Fonskiyonlar

### Map

`Map()` fonksiyonu, bir listedeki veya başka bir koleksiyondaki her elemana belirlediğin bir işlemi uygulamak için kullanılır.

```python
def divideNumber(number):
    return number / 2
```

```python
myList = [3,5,7,10,20,30]
```

```python
list(map(divideNumber,myList)) #myList içindeki her elemana divideNumber fonksiyonunu uygula demek.
```

### Filter

`Filter()` fonksiyonu, bir listedeki elemanlardan belirlediğimiz koşula uyanları seçmek için kullanılır.  
Yani `map()` bütün elemanlara işlem uygularken, `filter()` sadece koşulu sağlayan elemanları seçer.

```python
def controlString(string):
    return "Zehra" in string
```

```python
myStringList = ["Zehra", "Zehra Aydogdu","Aydogdu","Sultan"]
```

```python
list(filter(controlString,myStringList)) #Sadece içinde "Zehra" geçenleri bana gösterebiliyor.
```

### Lambda

Lambda, tek satırda kısa ve isimsiz fonksiyon (fonksiyon tanımlama) oluşturmamızı sağlar.  
Normalde bir fonksiyon yazmak için def kullanıyorduk. lambda ise özellikle kısa işlemlerde daha pratik bir yöntemdir

```python
lambdaislem = lambda sayi : sayi * 3
```

```python
lambdaislem(20)
```

```python
numList = [10,20,30,40,50]
```

```python
list(map(lambda num:num/4,numList)) #Anonim fonksiyon
```

### Normalizasyon Örneği

```python
data = [0,10, 20, 30, 40]

normalization = lambda x : (x-min(data))/(max(data)-min(data))
sonuc = list(map(normalization, data))

print(sonuc)
```
