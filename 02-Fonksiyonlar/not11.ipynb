# Recursive Fonksiyonlar

Rekürsif (özyineli) fonksiyonlar kendilerini çağırarak çalışırlar.  
Bir işlemin özyineli tanımlanabilmesi için temel bir sonlanma durumu mevcut olması gerekir

```python
def faktoriyel(n):
    if n == 0:
        return 1
    
    return n * faktoriyel(n-1)

print(faktoriyel(5))
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

for i in range(10):
    print(fibonacci(i))
```
