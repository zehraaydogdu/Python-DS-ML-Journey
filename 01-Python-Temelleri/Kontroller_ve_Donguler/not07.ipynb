## While Döngüsü

Bu olmaya devam ettiği sürece bunu yap. While döngüsü, belirli bir koşul doğru olduğu sürece kod bloğunu tekrar tekrar çalıştırır. Koşul yanlış olduğunda döngü sona erer.

```python
x = 0
```

```python
while x < 10:
    print(x)
    x += 1
```

```python
my_List = [10, 20, 30, 40, 50]
```

---

```python
#Değişkenleri print içinde yazdırmak istiyorsak:

print(f"My list: {my_List}")
```

```python
print("My List: ", my_List) #Başka bir yöntem de bu. Ama f-string yöntemi daha hızlı ve daha kolay.
```

```python
#Belli bir kapsamdaki sayıları yazdırmak istiyorum diyelim:
list(range(10))
```

```python
list(range(1, 10)) #1'den 10'a kadar olan sayıları yazdırır.
```

```python
list(range(1, 10, 2)) #1'den 10'a kadar olan tek sayıları yazdırır.
```

```python
#İndexleri yazdırmak istiyorsak:
for index in range(len(my_List)):
    print(f"Index: {index}")
```

```python
#Enumerate fonksiyonu ile hem indexleri hem de değerleri yazdırabiliriz:
for index, value in enumerate(my_List):
    print(f"Index: {index}, Value: {value}")
```
