## If Kontrolleri

```python
my_superhero = "WonderWoman"
```

```python
if my_superhero == "CaptainAmerica":
    print("Zehra")
elif my_superhero == "Batman":
    print("Batman")
else:
    print("Yok")
```

```python
my_superhero_list = ["Black Widow", "Spiderman", "WonderWoman"]
```

```python
if my_superhero in my_superhero_list:
    print("Bulundu")
```

## For Döngüsü

```python
my_List = [10, 20, 30, 40, 50]
```

```python
for buraya_ne_dedigim_onemsiz in my_List:
    print(buraya_ne_dedigim_onemsiz)
```

```python
for i in my_List:
    if i % 6 == 0:
        print(i)
```

## Continue & Break & Pass

```python
for x in my_List:
    print(x)
    if x == 30:
        print("Evreka")
        break #Döngüden çıkmak için kullanılır. Döngüden sonra kalan kodlar çalıştırılır.
```

```python
for sayi in my_List:
    print(sayi)
    if sayi == 30:
        print("Evreka")
        continue #Döngüdeki kalan kodları çalıştırmadan bir sonraki iterasyona geçmek için kullanılır.
        print("Devam")
```

```python
for sayim in my_List:
    pass #Burayı boş bırakmak istiyorum. Kod çalıştırılacak ama hiçbir şey yapılmayacak.
```
