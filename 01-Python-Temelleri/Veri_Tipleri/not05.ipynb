## Tuples

Birden fazla veriyi tek bir değişken içerisinde saklamamızı sağlayan veri türüdür.  
Listelere benzer ama aralarında çok önemli bir fark vardır: Tuple oluşturulduktan sonra elemanları değiştirilemez.

```python
myTuple = (10, "a", "b", 3.14)
```

```python
myTuple[0]
```

```python
myTuple[0] = 100 #Hata verir. Tuple'lar değiştirilemez.
```

## Veri Tipleri Karşılaştırması  
### (List, Tuple, Set, Dictionary)

| | `list` | `tuple` | `set` | `dict` |
| :--- | :--- | :--- | :--- | :--- |
| **Yazım** | `[1, 2]` | `(1, 2)` | `{1, 2}` | `{'a': 1}` |
| **Sıralı mı?** | Evet | Evet | Hayır | Ekleme sırası korunur |
| **Değiştirilebilir mi?** | Evet | Hayır | Evet | Evet |
| **Tekrar eden eleman** | Olabilir | Olabilir | Olamaz | Anahtarlar benzersiz |
| **Erişim** | İndeks: `L[0]` | İndeks: `T[0]` | Yok (`in` ile) | Anahtar: `D['a']` |
| **Boş oluşturma** | `[]` / `list()` | `()` / `tuple()` | `set()` | `{}` / `dict()` |

*Not: Python 3.7'den itibaren sözlükler eklenme sırasını korur.*

### Hangi Yapıyı Ne Zaman Kullanmalı?
