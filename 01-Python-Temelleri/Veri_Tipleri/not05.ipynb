## Tuples

Birden fazla veriyi tek bir değişken içerisinde saklamamızı sağlayan veri türüdür.  
Listelere benzer ama aralarında çok önemli bir fark vardır: Tuple oluşturulduktan sonra elemanları değiştirilemez.

```python
myTuple = (10, "a", "b", 3.14)
```

```python
myTuple[0]
```

```python
myTuple[0] = 100 #Hata verir. Tuple'lar değiştirilemez.
```
