## İndex

indexler bizim slotlarımız. İndexler hep 0dan başlıyor diğer kodlama dillerinde de olduğu gibi.  


````python
cumle[3] #cumle değişkeninin 3. indexindeki karakteri verir. (indexing)
````

````python
cumle[0:3] #cumle değişkeninin 0. ve 3. indexleri arasındaki karakterleri verir. (slicing)
````

````python
len(cumle) #cumle değişkeninin uzunluğunu verir. (length)
````

## Slicing

````python
# Üstte bahsettiğimiz cumle[0:3] örneği şunu anlatır: 
# cumle değişkeninin 0. indexinden 3. indexine kadar olan kısmı verir. (3. index dahil değil)
fullname[0:5] #fullname değişkeninin 0. indexinden 5. indexine kadar olan kısmı verir. (5. index dahil değil)
````

````python
fullname[3::] #fullname değişkeninin 3. indexinden sonuna kadar olan kısmı verir. (3. index dahil)
````

````python
fullname[:3:] #fullname değişkeninin başından 3. indexine kadar olan kısmı verir. (3. index dahil değil)
````

````python
fullname[::3] #fullname değişkeninin başından sonuna kadar olan kısmı 3'er 3'er atlayarak verir. (3. index dahil değil)
````

````python
fullname[3:5:]
fullname[3:5] #Aynı şeylerdir.
````

````python
fullname[::-1] #fullname değişkeninin başından sonuna kadar olan kısmı ters çevirerek verir
````
