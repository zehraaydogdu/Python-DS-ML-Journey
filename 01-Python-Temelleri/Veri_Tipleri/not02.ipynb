Pythonda bir şey yazarken onun yeşil olması o kelimenin pythonda tanımlı olduğu anlamına gelir.

Pythonda int / int işleminin sonucu floattır. Bölme işlemlerinin sonucunu float vermeyi tercih etmiştir.

````python
ilkSonuc = 2 ** 5 #2'nin 5. kuvvetini verir.
ikinciSonuc = 10 % 3 #10'u 3'e böldüğümüzde kalanı verir. (remainder)
print(ilkSonuc)
print(ikinciSonuc)
````

## İndex

indexler bizim slotlarımız. İndexler hep 0dan başlıyor diğer kodlama dillerinde de olduğu gibi.  


````python
cumle[3] #cumle değişkeninin 3. indexindeki karakteri verir. (indexing)
print()
cumle[0:3] #cumle değişkeninin 0. ve 3. indexleri arasındaki karakterleri verir. (slicing)
````
