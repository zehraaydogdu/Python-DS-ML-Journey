- Sigmoid Fonksiyonu: Sigmoid, aldığı herhangi bir sayıyı 0 ile 1 arasındaki bir değere dönüştüren matematiksel fonksiyondur.

```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))
for x in [-2, 0, 2]: #Fonksiyonumuza sırasıyla -2, 0 ve 2 değerlerini verip çıktısını alıyoruz.
    print(x, round(sigmoid(x), 4))
```

### Random Kütüphanesi

Rastgele sayılar üretmek, listelerden eleman seçmek, listeyi karıştırmak ve rastgele veri oluşturmak için kullanılır.

```python
import random

random.seed(7) #Normalde random her çalıştırıldığında farklı sonuçlar üretir. Seed yani tohum verdik ki aynı sonucu üretsin.
print(random.random()) #0 ile 1 arasında rastgele bir ondalıklı sayı üretir.
print(random.uniform(1.5, 3.5)) #Belirttiğimiz aralıkta rastgele bir ondalıklı sayı üretir.
print(random.randint(1, 6)) #Belirttiğimiz aralıkta rastgele bir tam sayı üretir. 1 ve 6 dahil.
print(random.randrange(0, 100, 5)) #Belirttiğimiz aralıkta ve adımda rastgele bir tam sayı üretir. 0 ve 100 dahil değildir.
#random.randrange(baslangic, bitis, adim) bitiş dahil değildir.
```

```python
#Diyelim ki bir listemiz var:
renkler = ["kirmizi", "mavi", "yeşil", "sari"]
print(random.choice(renkler)) #Listemizden rastgele bir eleman seçer.
print(random.choices(renkler, k=3)) #Listemizden rastgele 3 eleman seçer. Choicede aynı eleman birden fazla seçilebilir. Tekrarlı.

print(random.sample(renkler, 2)) #Listemizden rastgele 2 eleman seçer. Sampleda aynı eleman birden fazla seçilemez. Tekrarsız.
```

```python
kartlar = ["A", "K", "Q", "J", "10"]  
random.shuffle(kartlar) #Kartları karıştırdık.
print(kartlar) 
```

```python
sayilar = [1, 2, 3, 4, 5]
yeni = random.sample(sayilar, len(sayilar)) #Orijinal listemizi korumak için.

print(sayilar)
print(yeni)
```

```python
import random

random.seed(10)
boylar = [
    round(random.gauss(170, 8), 1) #Normal dağılımdan rastgele boylar üretir. 170 ortalama, 8 standart sapma.
    for _ in range(6) #Biz kaçıncı atış olduğunu önemsemeden 6 kez atış yapıyoruz. Bu yüzden i değil alt çizgi koyduk.
]

print(boylar)
```

```python
sonuc = random.choices( #Bu da ağırlıklı seçim. Biz 70% olasılıkla yazı, 30% olasılıkla tura gelmesini istiyoruz.
    ["Yazi", "Tura"],
    weights=[70, 30],
    k=10
)
print(sonuc.count("Yazi"), "yazi geldi") #Çıkan sonuç örn 7 ise 10 atışta 7 kez yazı geldi demektir.
```

```python
import random

veri = list(range(1, 21))
random.seed(42)
random.shuffle(veri)
sinir = int(len(veri) * 0.8)

egitim, test = veri[:sinir], veri[sinir:] #Eğitim ve test verilerini ayırdık. Eğitim verisi 1-16 arası, test verisi 17-20 arası olacak şekilde.

print("Eğitim:", len(egitim), "örnek")
print("Test  :", test) 
```

```python
import string

metin = "Merhaba, Dünya! Python ÇOK güzel... #yapayzeka"
tablo = str.maketrans("","", string.punctuation) #str.maketrans(eski, yeni, silinecek). Değiştirilecek, yerine gelecek, silinecek.
temiz = metin.translate(tablo).lower() #Hazırladığımız tabloyu metne uygular.
print(temiz)

kelimeler = temiz.split() #Metni boşluklardan ayırıp listeye dönüştürür.
print(kelimeler)
print(len(kelimeler), "kelime")
```

```python
import string

def sifre_guclu_mu(s):
    return (len(s) >= 8
            and any(c in string.ascii_uppercase for c in s) #Any fonksiyonu: Bu koşulu sağlayan en az bir karakter var mı ?
            and any(c in string.digits for c in s)
            and any(c in string.punctuation for c in s))

for s in ["python123", "Python123", "Pyth0n!2026"]:
    print(s, "->", sifre_guclu_mu(s))
```

### Re Kütüphanesi

Re metin içinde belirli kalıpları bulmamızı sağlayan kütüphanedir.

\d	Rakam  
\w	Harf, rakam veya _  
\s	Boşluk karakteri  
.	Herhangi bir karakter  
[abc]	a, b veya c  
^	Metnin başlangıcı  
$	Metnin sonu  
Bir veya daha fazla  +  
Sıfır veya daha fazla  *  
Sıfır veya bir  ?  
{n}	Tam n kez  
{n,m}	n ile m arasında  
\.	Gerçek nokta işareti

```python
import re

metin = "Sipariş no: 4521, tutar: 349 TL, tarih: 19.10.2026"
print(re.search(r"\d+", metin).group()) # ilk eşleşen sayı
print(re.findall(r"\d+", metin)) #Metindeki bütün eşleşmeleri bulur.
print(re.findall(r"\d{2}\.\d{2}\.\d{4}", metin)) #Tarihi bulur.
print(bool(re.match(r"Sipariş", metin))) #Metin Sipariş ile başlıyorsa True döner.
```

### Küme İşlemleri (Sets)

Matematikteki kümelerin bazı karşılıkları:

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
 
print(A | B)     # birleşim
print(A & B)     # kesişim
print(A - B)     # fark
print(B - A)
print(A ^ B)     # simetrik fark
 
print(A.union(B)) #Birleşim işlemi yapar. A ve B kümelerinin birleşimini verir.
print(A.intersection(B)) #Kesişim işlemi yapar. A ve B kümelerinin kesişimini verir.
```
