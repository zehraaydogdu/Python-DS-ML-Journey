## Listeler

Listeler bir veri tipi değil bir veri yapısıdır. İçinde birden fazla veri tutar.   


String ifadelerimizin tek bir elemanını değiştiremeyiz. Tamamı olur ama bir eleman olmaz.  
Mesela myString[0] = 'a' yapamayız. Buna immutable ya da immutability denir.


```python
myList = [10 , 5, 8, 3]
```

```python
myList.append(60) #Listeye yeni bir eleman ekler.
```

```python
myList
```

```python
myList.extend([70, 80]) #Listeye elemanları ayrı ayrı ekler.
```

```python
myList 
```

```python
myList.append([90, 100]) #Listeye elemanları tek bir liste olarak ekler.
```

```python
myList 
```

```python
print(myList.count(60)) #Belirtilen değerin listede kaç kez tekrarlandığını yazdırır.
```

```python
myList.insert(3, 100) #Belirtilen indexe yeni bir eleman ekler.
```

```python
myList
```

```python
myList.remove(5)
print(myList) #Belirtilen değeri kaldırır.
```

```python
myList.pop() #Son elemanı kaldırır.
print(myList)
```

```python
myList.reverse() #Listeyi tersine çevirir.
myList
```

```python
myList.sort() #Listeyi küçükten büyüğe sıralar.
myList
```

```python
print(myList.copy()) #Listeyi kopyalar.
```
